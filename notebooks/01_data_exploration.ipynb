```python
import pandas as pd
import numpy as np

print("Python environment is working!")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
# Load the raw athlete dataset

df = pd.read_csv("../data/football_athlete_sessions_raw.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Basic dataset inspection

print("First 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Check important numerical ranges

numeric_columns = [
    "Age",
    "Duration_Min",
    "Distance_Km",
    "Sprint_Distance_Km",
    "Intensity_Score",
    "Avg_Heart_Rate",
    "Max_Heart_Rate",
    "Sleep_Hours",
    "Recovery_Score",
    "Goals",
    "Assists",
    "Pass_Accuracy",
    "Performance_Score"
]

df[numeric_columns].describe().T
```

```python
# Convert Date from text to datetime

df["Date"] = pd.to_datetime(df["Date"])

print("Date conversion successful!")
print("Date type:", df["Date"].dtype)
print("Date range:", df["Date"].min(), "to", df["Date"].max())
```

```python
# Calculate daily training workload

df["Daily_Workload"] = (
    df["Duration_Min"] * df["Intensity_Score"]
).round(2)

print("Daily workload calculated successfully!")
display(
    df[
        ["Athlete_ID", "Date", "Duration_Min",
         "Intensity_Score", "Daily_Workload"]
    ].head(10)
)
```

```python
# Make sure records are sorted correctly
df = df.sort_values(["Athlete_ID", "Date"]).reset_index(drop=True)

# Calculate rolling workload for each athlete
df["Rolling_7D_Workload"] = (
    df.groupby("Athlete_ID", group_keys=False)
      .apply(
          lambda group: group.set_index("Date")["Daily_Workload"]
          .rolling("7D")
          .sum()
          .set_axis(group.index)
      )
      .sort_index()
)

df["Rolling_28D_Workload"] = (
    df.groupby("Athlete_ID", group_keys=False)
      .apply(
          lambda group: group.set_index("Date")["Daily_Workload"]
          .rolling("28D")
          .sum()
          .set_axis(group.index)
      )
      .sort_index()
)

print("Rolling workload calculated successfully!")

display(
    df[
        [
            "Athlete_ID",
            "Date",
            "Daily_Workload",
            "Rolling_7D_Workload",
            "Rolling_28D_Workload"
        ]
    ].head(10)
)
```

```python
# Calculate each athlete's personalized workload baseline
# using previous 28-day workload observations

df["Personal_Baseline"] = (
    df.groupby("Athlete_ID")["Rolling_28D_Workload"]
      .transform(
          lambda x: x.shift(1).rolling(window=10, min_periods=3).mean()
      )
)

print("Personalized baseline calculated successfully!")

display(
    df[
        [
            "Athlete_ID",
            "Date",
            "Rolling_7D_Workload",
            "Rolling_28D_Workload",
            "Personal_Baseline"
        ]
    ].head(15)
)
```

```python
# Calculate workload deviation from the athlete's personal baseline

df["Workload_Deviation"] = (
    (df["Rolling_7D_Workload"] - df["Personal_Baseline"])
    / df["Personal_Baseline"]
) * 100

df["Workload_Deviation"] = df["Workload_Deviation"].round(2)

print("Workload deviation calculated successfully!")

display(
    df[
        [
            "Athlete_ID",
            "Date",
            "Rolling_7D_Workload",
            "Personal_Baseline",
            "Workload_Deviation"
        ]
    ].head(20)
)
```

```python
# Detect workload anomalies

df["Anomaly_Flag"] = np.where(
    df["Workload_Deviation"].abs() >= 25,
    "Anomaly",
    "Normal"
)

print("Anomaly detection completed!")

print("\nAnomaly distribution:")
print(df["Anomaly_Flag"].value_counts())

display(
    df[
        [
            "Athlete_ID",
            "Date",
            "Rolling_7D_Workload",
            "Personal_Baseline",
            "Workload_Deviation",
            "Anomaly_Flag"
        ]
    ].tail(20)
)
```

```python
# Calculate athlete readiness score

workload_balance = (
    100 - df["Workload_Deviation"].abs().clip(upper=100)
)

sleep_score = (
    df["Sleep_Hours"] / 9 * 100
).clip(upper=100)

df["Readiness_Score"] = (
    0.35 * df["Recovery_Score"]
    + 0.25 * df["Performance_Score"]
    + 0.20 * workload_balance
    + 0.20 * sleep_score
).clip(0, 100).round(2)

print("Readiness score calculated successfully!")

display(
    df[
        [
            "Athlete_ID",
            "Date",
            "Recovery_Score",
            "Sleep_Hours",
            "Workload_Deviation",
            "Performance_Score",
            "Readiness_Score"
        ]
    ].tail(15)
)
```

```python
# Create the target: next recorded performance score for each athlete

df = df.sort_values(["Athlete_ID", "Date"]).reset_index(drop=True)

df["Next_Performance_Score"] = (
    df.groupby("Athlete_ID")["Performance_Score"].shift(-1)
)

print("Prediction target created!")

display(
    df[
        [
            "Athlete_ID",
            "Date",
            "Performance_Score",
            "Next_Performance_Score"
        ]
    ].head(15)
)
```

```python
# Prepare data for machine learning

ml_df = df.dropna(subset=["Next_Performance_Score"]).copy()

features = [
    "Age",
    "Duration_Min",
    "Distance_Km",
    "Sprint_Distance_Km",
    "Intensity_Score",
    "Avg_Heart_Rate",
    "Max_Heart_Rate",
    "Sleep_Hours",
    "Recovery_Score",
    "Goals",
    "Assists",
    "Pass_Accuracy",
    "Daily_Workload",
    "Rolling_7D_Workload",
    "Rolling_28D_Workload",
    "Workload_Deviation"
]

X = ml_df[features]
y = ml_df["Next_Performance_Score"]

print("ML dataset prepared!")
print("Rows:", len(ml_df))
print("Features:", len(features))
print("Target:", "Next_Performance_Score")
```

```python
# Time-based train/test split

split_date = ml_df["Date"].quantile(0.80)

train_df = ml_df[ml_df["Date"] <= split_date].copy()
test_df = ml_df[ml_df["Date"] > split_date].copy()

X_train = train_df[features]
y_train = train_df["Next_Performance_Score"]

X_test = test_df[features]
y_test = test_df["Next_Performance_Score"]

print("Time-based split completed!")
print("Split date:", split_date)
print("Training records:", len(train_df))
print("Testing records:", len(test_df))
print("Training period:", train_df["Date"].min(), "to", train_df["Date"].max())
print("Testing period:", test_df["Date"].min(), "to", test_df["Date"].max())
```

```python
# Baseline model: Linear Regression

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

baseline_model = LinearRegression()

baseline_model.fit(X_train, y_train)

baseline_predictions = baseline_model.predict(X_test)

baseline_mae = mean_absolute_error(y_test, baseline_predictions)
baseline_rmse = np.sqrt(mean_squared_error(y_test, baseline_predictions))
baseline_r2 = r2_score(y_test, baseline_predictions)

print("Baseline Model: Linear Regression")
print("----------------------------------")
print(f"MAE  : {baseline_mae:.2f}")
print(f"RMSE : {baseline_rmse:.2f}")
print(f"R²   : {baseline_r2:.3f}")
```

```python
# Remove rows with missing values in any ML feature or target

ml_df = df.dropna(
    subset=features + ["Next_Performance_Score"]
).copy()

X = ml_df[features]
y = ml_df["Next_Performance_Score"]

# Recreate the time-based split after removing incomplete rows
split_date = ml_df["Date"].quantile(0.80)

train_df = ml_df[ml_df["Date"] <= split_date].copy()
test_df = ml_df[ml_df["Date"] > split_date].copy()

X_train = train_df[features]
y_train = train_df["Next_Performance_Score"]

X_test = test_df[features]
y_test = test_df["Next_Performance_Score"]

print("Clean ML dataset prepared!")
print("Usable records:", len(ml_df))
print("Training records:", len(train_df))
print("Testing records:", len(test_df))
print("Remaining missing values:", X_train.isna().sum().sum())
```

```python
# Baseline model: Linear Regression

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

baseline_model = LinearRegression()

baseline_model.fit(X_train, y_train)

baseline_predictions = baseline_model.predict(X_test)

baseline_mae = mean_absolute_error(y_test, baseline_predictions)
baseline_rmse = np.sqrt(mean_squared_error(y_test, baseline_predictions))
baseline_r2 = r2_score(y_test, baseline_predictions)

print("Baseline Model: Linear Regression")
print("----------------------------------")
print(f"MAE  : {baseline_mae:.2f}")
print(f"RMSE : {baseline_rmse:.2f}")
print(f"R²   : {baseline_r2:.3f}")
```

```python
# Advanced model: XGBoost

from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

xgb_model.fit(X_train, y_train)

xgb_predictions = xgb_model.predict(X_test)

xgb_mae = mean_absolute_error(y_test, xgb_predictions)
xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_predictions))
xgb_r2 = r2_score(y_test, xgb_predictions)

print("Advanced Model: XGBoost")
print("------------------------")
print(f"MAE  : {xgb_mae:.2f}")
print(f"RMSE : {xgb_rmse:.2f}")
print(f"R²   : {xgb_r2:.3f}")
```

```python
# Compare baseline and advanced models

comparison = pd.DataFrame({
    "Model": ["Linear Regression", "XGBoost"],
    "MAE": [baseline_mae, xgb_mae],
    "RMSE": [baseline_rmse, xgb_rmse],
    "R2": [baseline_r2, xgb_r2]
})

display(comparison)
```

```python
# Analyze prediction errors

results = test_df[
    ["Athlete_ID", "Date", "Performance_Score", "Next_Performance_Score"]
].copy()

results["Linear_Prediction"] = baseline_predictions
results["XGBoost_Prediction"] = xgb_predictions

results["Linear_Absolute_Error"] = (
    results["Next_Performance_Score"] - results["Linear_Prediction"]
).abs()

results["XGBoost_Absolute_Error"] = (
    results["Next_Performance_Score"] - results["XGBoost_Prediction"]
).abs()

display(results.head(15))

print("Average Linear Regression error:",
      results["Linear_Absolute_Error"].mean().round(2))

print("Average XGBoost error:",
      results["XGBoost_Absolute_Error"].mean().round(2))
```

```python
# Personalized athlete workload baseline

df = df.sort_values(["Athlete_ID", "Date"]).copy()

# Previous workload for each athlete
df["Previous_7D_Workload"] = (
    df.groupby("Athlete_ID")["Rolling_7D_Workload"]
      .shift(1)
)

# Personalized baseline = average of previous workload values
df["Personal_Workload_Baseline"] = (
    df.groupby("Athlete_ID")["Previous_7D_Workload"]
      .transform(lambda x: x.expanding().mean())
)

# Percentage deviation from personal baseline
df["Workload_Deviation_Pct"] = (
    (df["Rolling_7D_Workload"] - df["Personal_Workload_Baseline"])
    / df["Personal_Workload_Baseline"]
) * 100

print("Personalized baseline calculated successfully!")

display(
    df[
        [
            "Athlete_ID",
            "Date",
            "Rolling_7D_Workload",
            "Personal_Workload_Baseline",
            "Workload_Deviation_Pct"
        ]
    ].head(15)
)
```

```python
# Workload anomaly detection

df["Workload_Anomaly"] = (
    df["Workload_Deviation_Pct"].abs() > 20
)

df["Anomaly_Status"] = df["Workload_Anomaly"].map({
    True: "⚠️ Anomaly",
    False: "Normal"
})

print("Workload anomaly detection completed!")

display(
    df[
        [
            "Athlete_ID",
            "Date",
            "Rolling_7D_Workload",
            "Personal_Workload_Baseline",
            "Workload_Deviation_Pct",
            "Anomaly_Status"
        ]
    ].head(20)
)

print("\nTotal anomaly records:",
      df["Workload_Anomaly"].sum())
```

```python
print("Total records:", len(df))
print("Normal records:", (~df["Workload_Anomaly"]).sum())
print("Anomaly records:", df["Workload_Anomaly"].sum())

print("\nAnomaly percentage:",
      round(df["Workload_Anomaly"].mean() * 100, 2), "%")

print("\nSample anomalies:")

display(
    df[df["Workload_Anomaly"]][
        [
            "Athlete_ID",
            "Date",
            "Rolling_7D_Workload",
            "Personal_Workload_Baseline",
            "Workload_Deviation_Pct",
            "Anomaly_Status"
        ]
    ].head(15)
)
```

```python
# Improved personalized anomaly detection
# Use the athlete's previous 28 days as their personal baseline

df = df.sort_values(["Athlete_ID", "Date"]).copy()

# Previous 28-day workload baseline
df["Personal_28D_Baseline"] = (
    df.groupby("Athlete_ID")["Rolling_7D_Workload"]
      .transform(lambda x: x.shift(1).rolling(28, min_periods=7).mean())
)

# Personal workload standard deviation
df["Personal_28D_Std"] = (
    df.groupby("Athlete_ID")["Rolling_7D_Workload"]
      .transform(lambda x: x.shift(1).rolling(28, min_periods=7).std())
)

# Z-score: how unusual is the current workload for this athlete?
df["Workload_Z_Score"] = (
    (df["Rolling_7D_Workload"] - df["Personal_28D_Baseline"])
    / df["Personal_28D_Std"].replace(0, np.nan)
)

# Flag unusually high/low workload
df["Workload_Anomaly_Improved"] = (
    df["Workload_Z_Score"].abs() > 2
)

df["Anomaly_Status_Improved"] = df["Workload_Anomaly_Improved"].map({
    True: "⚠️ Anomaly",
    False: "Normal"
})

print("Improved personalized anomaly detection completed!")

print(
    "Anomaly records:",
    df["Workload_Anomaly_Improved"].sum()
)

print(
    "Anomaly percentage:",
    round(
        df["Workload_Anomaly_Improved"].mean() * 100,
        2
    ),
    "%"
)

display(
    df[
        [
            "Athlete_ID",
            "Date",
            "Rolling_7D_Workload",
            "Personal_28D_Baseline",
            "Workload_Z_Score",
            "Anomaly_Status_Improved"
        ]
    ].dropna().head(20)
)
```
